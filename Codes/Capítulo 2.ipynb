### Exemplo 2.2
Um gás encontra-se confinado em um cilindro por um êmbolo. A pressão inicial do gás é de $7\text{ bar}$, e o seu volume é de $0,10\text{ m}^3$. O êmbolo é mantido imóvel por presilhas localizadas na parede do cilindro. O equipamento completo encontra-se no vácuo total. Qual é a variação de energia do equipamento, se as presilhas forem removidas de tal forma que o gás se expanda subitamente para o dobro do seu volume inicial, com o êmbolo sendo retido por outras presilhas no final do processo?

---

Como o enunciado considera o equipamento como um todo, o sistema é definido pelo conjunto formado pelo gás, o êmbolo e o cilindro. Durante o processo, não há realização de trabalho, pois não existe atuação de forças externas sobre o sistema, e também não ocorre troca de calor, uma vez que o vácuo ao redor impede a transferência térmica. Dessa forma, tanto o calor 
𝑄 quanto o trabalho 𝑊 são nulos, o que implica que a energia total do sistema se conserva. Entretanto, na falta de informações adicionais, não é possível determinar como essa energia está distribuída entre os diferentes componentes do sistema, sendo possível que essa distribuição seja distinta da condição inicial.

### Exemplo 2.3
Se o processo descrito no Exemplo 2.2 for repetido, não no vácuo, mas no ar, a uma pressão atmosférica padrão de $101,3\text{ kPa}$, qual é a variação de energia do equipamento? Admita que a taxa de transferência de calor entre o equipamento e o ar ambiente seja lenta quando comparada à taxa na qual o processo ocorre.

---

```python
Patm = 101.3 #[kPa]
Vf = 0.2 #[m³]
Vi = 0.1 #[m³]

W = -Patm*(Vf-Vi) #[kJ] 
Q = 0 
dEsist = Q + W



print(f"A variação de Energia do equipamento é: {round(dEsist, 2)} kJ")
```

### Exemplo 2.4
Quando um sistema é levado do estado $a$ para o estado $b$ na Figura 2.1, ao longo da trajetória $acb$, $100\text{ J}$ de calor fluem para dentro do sistema e o sistema realiza $40\text{ J}$ de trabalho.



(a) Qual a quantidade de calor que flui para dentro do sistema ao longo da trajetória $aeb$, se o trabalho realizado pelo sistema for de $20\text{ J}$?  
(b) O sistema retorna de $b$ para $a$ pela trajetória $bda$. Se o trabalho realizado sobre o sistema for de $30\text{ J}$, o sistema absorverá ou liberará calor? Qual a quantidade?

---

```python
#a) 
Qacb = 100 #[J]
Wacb = -40 #[J]
Waeb = -20 #[J]

dUab = Qacb + Wacb 
Qaeb = dUab - Waeb 


#b) 
Wbda = 30
dUba = - dUab 
Qbda = dUab - Wbda

print(f"A) A quantidade de calor é: {round(Qaeb)} kJ") 
print(f"B) O sistema liberará calor e a quantidade é: {round(Qbda)} kJ") 
```

### Exemplo 2.6
Um dispositivo horizontal cilindro/êmbolo é colocado em um banho termostático. O êmbolo desliza no cilindro com atrito desprezível, e uma força externa o mantém em posição contra uma pressão inicial do gás de $14\text{ bar}$. O volume inicial do gás é de $0,03\text{ m}^3$. A força externa sobre o êmbolo é gradualmente reduzida, permitindo a expansão isotérmica do gás até que o seu volume dobre. Se o volume do gás estiver relacionado com a sua pressão de tal forma que o produto $PV$ seja constante, qual o trabalho realizado pelo gás ao deslocar a força externa?

Que quantidade de trabalho seria realizado, se a força externa fosse subitamente reduzida à metade do seu valor inicial em vez de ser gradualmente reduzida?

---


```python
import math as mt 
#Primeiro caso:
Pi = 14*1e5 #[bar] 
Vi = 0.03 #[m³]
Vf = 0.06 #[m³]
k = Pi*Vi 
W = -k*mt.log(Vf/Vi) 
Pf = k/Vf  

#Segundo caso: Calculando o trabalho realizado se a força externa fosse subitamente reduzida à metade 
W2 = - (Pi/2)*(Vf-Vi)

#Calculando a Eficiência:
ef = (W2/W)

print(f"O valor do trabalho realizado é de:{round(W)} J")
print(f"O valor da pressão final é de: {round(Pf)} Pa")
print(f"O valor do trabalho realizado com a força sendo reduzida pela metade é de:{round(W2)} J")
print(f"A eficiência do W2 em relação ao W é de: {round(ef, 3)}")
```
