Como o enunciado considera o equipamento como um todo, o sistema é definido pelo conjunto formado pelo gás, o êmbolo e o cilindro. Durante o processo, não há realização de trabalho, pois não existe atuação de forças externas sobre o sistema, e também não ocorre troca de calor, uma vez que o vácuo ao redor impede a transferência térmica. Dessa forma, tanto o calor 
𝑄 quanto o trabalho 𝑊 são nulos, o que implica que a energia total do sistema se conserva. Entretanto, na falta de informações adicionais, não é possível determinar como essa energia está distribuída entre os diferentes componentes do sistema, sendo possível que essa distribuição seja distinta da condição inicial.
